```python
import pymongo
import pandas as pd
import pandas_datareader as pdr
```

```python
def open_db(db_name):
    client = pymongo.MongoClient("mongodb://localhost:27017/")
    db = client[db_name]
    return db
```

```python
db = open_db('Stocks')
docs = db.US_Stocks.find({"bscs.symbol":"AAPL"})
doc = docs.next()
```
