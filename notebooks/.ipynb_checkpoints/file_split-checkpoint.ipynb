```python
import os
import math

import re

def atoi(text):
    return int(text) if text.isdigit() else text

def natural_keys(text):
    '''
    alist.sort(key=natural_keys) sorts in human order
    http://nedbatchelder.com/blog/200712/human_sorting.html
    (See Toothy's implementation in the comments)
    '''
    return [ atoi(c) for c in re.split(r'(\d+)', text) ]

db_file="/home/vpetla/work/stockanalysis/db_backup/Stocks/US_Stocks.bson"
max_size = 50*1024*1024 # 50MB

def split(db_file, max_size, store_path, splits_name):
    f = open(db_file,"rb")
    size = os.path.getsize(db_file)
    print(size)
    splits = math.ceil(size/max_size)
    print(splits)

    for i in range(splits):
        filex = "/home/vpetla/work/stockanalysis/db_backup/Stocks/US_Stocks/US_Stocks.bson%r" %(i)
        print(filex)
        f.seek(i*max_size)
        buf = f.read(max_size)
        fx = open(filex, "wb")
        fx.write(buf)
        fx.close()

def combine(store_path, combine_name):
    db_file = "%r/%r" %(store_path, combine_name)
    #f = open(db_file, "ab+")
    with open(db_file, 'ab+') as f:
        print(f.tell())
        for (root,dirs,files) in os.walk(store_path, topdown=True):
            files.sort(key=natural_keys)
            for fx in files:
                print(fx)
                fxb = open(fx, "r")
                with open(fx, 'r') as fxb:
                    f.write(fxb.read())
        
combine("/home/vpetla/work/stockanalysis/db_backup/Stocks/US_Stocks", "US_Stocks.bson")
```
