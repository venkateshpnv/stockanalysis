```python
from datetime import datetime as dt

val="86. Wednesday, Jun 19, 2019, 58.356."
price=float(val.split(",")[-1][:-1])
print(price)
date=val.rsplit(",",1)[0].split(".")[-1].split(",", 1)[-1].lstrip().rstrip().replace(",","")
print(date)
date=dt.strptime(date, '%b %d %Y').date()
str(date)
#date.strftime("%B %d, %Y")
```
