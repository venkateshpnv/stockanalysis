```python
key_value = {}
my_dictionary = {}
#Taking dynamic key value pair
list_key = ["a","b","c"]
list_val = ["A","B","C"]
my_dic_input = {"a":"A","b":"B","c":"C"}
my_dict_output = {}
#Printing the length of the key list
lenList = len(list_key)
#print(lenList)
#Method 1
#Iterating for lenList times
for elements in range(0,lenList) :
	#Extracting the value of key and value at the particular positions
	key = list_key[elements]
	value = list_val[elements]
	#Storing the extracted key value pair in the dictionary
	key_value[key]['col'] = value
#Method 2
for key, value in my_dic_input.items():
	#print(('%s-%s')% (key, value))
	my_dict_output['%s' %(key)]['col'] = '%s' % (value)
#printing the result
print("------Result through method 1------------")
print(key_value)
print("-------Result through method 2---------------")
print(my_dict_output)
```
